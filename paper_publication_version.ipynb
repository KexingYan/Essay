## Abstract

This paper studies how Kiva microfinance lending activity relates to economic development across an
84-country panel spanning 2013–2017. Using a country-year panel (357 observations), a loan-level
panel of individual Kiva loans, and a global country-year panel that adds true zero-lending
observations, I test whether lending activity follows the inverted-U ("middle-income peak") pattern
suggested by early descriptive plots — and, if so, at which of three margins. Conditional on Kiva
already lending in a country, total lending volume does not show a significant inverted-U: a
quadratic term in log GDP per capita is negative in every country-year specification but never
statistically distinguishable from zero (p > 0.10), a conclusion confirmed by a formal Lind–Mehlum
(2010) U-test and unchanged whether estimated by OLS-in-logs or by PPML on a global sample including
zeros. At the loan level, a quadratic term on individual loan size is positive and highly
significant (p < 0.01) — a U-shape, not an inverted-U — that survives excluding the US and sector
fixed effects, and that attenuates (from 0.245 to 0.148) but remains significant when loan size is
measured relative to local income rather than in nominal dollars. At the **extensive** margin —
whether Kiva operates in a country at all, estimated on a global panel of 1,020 country-years
including 646 zero-lending observations — a Logit model finds significant concavity: the quadratic
term is negative and significant (p = 0.023), and entry probability declines steeply with
development, from roughly two-thirds of country-years in the poorest deciles to below 7% in the
richest three. Held to the same Lind–Mehlum standard as the other margins, however, the initial
*rise* in entry probability at the very bottom of the income distribution is not statistically
established (low-end slope t = 1.07, one-sided p = 0.143), so the extensive margin is best described
as concave and sharply declining, with suggestive but unconfirmed evidence of a peak just above the
lowest observed incomes. The three results are not in tension once the outcomes are understood as
measuring different things: **whether a country receives any Kiva lending at all** (extensive
margin, concave and declining), **how much total lending a country receives conditional on receiving
any** (quantity margin, not nonlinear), and **how large a typical loan is** (size margin, U-shaped).
The paper's contribution is documenting this three-way divergence and holding every margin to the
same formal shape test rather than reading significance off a single quadratic coefficient: once
that discipline is applied, the only confirmed nonlinearity is the loan-level U-shape in loan size,
and the original "middle-income peak" intuition survives only as significant concavity — a steep
decline with an unconfirmed initial rise — at the extensive margin.

## 1. Introduction

Microfinance is often described as a tool for expanding credit access where formal banking systems are
thin, but the empirical literature does not support a simple view that more poverty automatically implies
more successful microfinance. Early survey and review work emphasizes both the promise and the limits of
microfinance, especially the tension between outreach to poor borrowers and the operational realities of
lending in weak institutional environments (Morduch, 1999). A second strand of work studies microfinance
institutions directly and argues that financial sustainability, commercialization, and outreach often move
together only imperfectly. Cull, Demirgüç-Kunt, and Morduch (2007, 2009) show that institutions can scale,
but doing so may involve trade-offs between profitability and reaching poorer clients. Hermes, Lensink,
and Meesters (2011) similarly find that deeper outreach can be associated with lower operating efficiency.

A third strand asks whether microcredit actually improves welfare for poor households. Some influential
studies report gains in access to credit and business activity but more limited average effects on
broader welfare outcomes. Banerjee, Duflo, Glennerster, and Kinnan (2015), Angelucci, Karlan, and Zinman
(2015), Tarozzi, Desai, and Johnson (2015), and Karlan and Zinman (2011) all contribute to a more cautious
consensus: microcredit may help some borrowers and relax constraints, but average poverty effects are not
universally large or immediate. This more skeptical perspective is reinforced by replication and
reanalysis work such as Roodman and Morduch (2014), which highlights how sensitive headline claims can be
to identification choices and model specification. At the same time, classic work on financial access and
poverty, such as Pitt and Khandker (1998) and Burgess and Pande (2005), keeps attention on the idea that
credit access can matter, especially when it expands into underserved places.

These papers are highly relevant, but most focus on either household-level treatment effects or
institution-level performance. This paper asks a different question at the country-year and loan level:
**where is platform-based microfinance most active across countries, and does the answer depend on whether
"activity" is measured by whether Kiva operates there at all, by total volume, or by the size of individual
loans?** Country-level Kiva lending data capture the geographic allocation of microfinance activity rather
than only average borrower outcomes or the internal efficiency of a single lender; a global panel that adds
zero-lending country-years lets the same question be asked about the *extensive* margin (does Kiva operate
in a country at all); and the loan-level data let it be asked about the *intensive* margin — how large a
loan Kiva borrowers receive, conditional on receiving one.

The paper uses an **X → Y → Z** framework, but with three versions of **Y**. The first, **Kiva presence**
(`kiva_active`), captures whether Kiva lends in a country-year at all. The second, **total lending volume**
(`log_total_loan_amount`), captures how much Kiva activity a country receives conditional on Kiva being
present. The third, **individual loan size** (`log_loan_amount`), captures how large a typical loan is once
a loan has been made. The main explanatory variable **X** is economic development, proxied by log GDP per
capita. The mechanism **Z** is institutional quality and financial access, which may shape whether
development translates into more or less microfinance activity. Kiva lending data are combined with World
Bank macroeconomic indicators, governance variables, and country-level population measures to build all
three panels.

Initial descriptive plots (Section 3) suggest a possible middle-income hump in total lending volume,
which motivated an original hypothesis that lending follows an inverted-U pattern peaking in
lower-middle income countries. That hypothesis does not survive formal regression testing at the
country level (Section 5.1) conditional on Kiva already operating there: the quadratic term is never
significant, a conclusion confirmed by a formal U-test (Section 5.4). What the data *do* support,
robustly, is a divergence across margins: whether Kiva enters a country at all is significantly
concave in development — entry probability falls steeply with income, though the initial rise among
the very poorest does not pass the formal low-end slope test (Sections 5.3–5.4); total lending
volume conditional on entry trends flat-to-downward with development; and the size of individual
loans follows a significant U-shape — falling and then rising with income (Section 5.2). Section 5
develops all three results; Section 6 checks robustness; Section 7 discusses limitations, including
that the size and quantity margins remain descriptive, non-causal analyses of countries and loans
that Kiva has already chosen to serve.

## 2. Data

The central empirical question is whether Kiva lending activity is concentrated in the poorest countries,
clusters at intermediate development levels, or instead varies with development in some other way — and
whether that answer changes depending on whether "activity" means total volume or typical loan size. To
answer it, I construct a country-year panel and a loan-level panel, define the main X, Y, and Z variables,
and compare lending patterns across countries with different levels of development.

### 2.1 Construction

The main dataset begins with Kiva loan-level information and aggregates it to the **country-year** level
for the period **2013–2017**. I sum loan amounts and loan counts within each country-year and then merge
those aggregates with country-level development and institutional variables. The final country-year
analysis file is therefore a merged panel rather than a pure Kiva extract; a second, loan-level file keeps
each individual loan as its own observation and merges in the same country-year covariates (Section 5.2).

The country-year panel contains the following main variables:

- **Y (quantity): Kiva lending activity**, measured by **log_total_loan_amount** and, for descriptive
  mapping only, **loan_per_100k**.
- **Y (size): individual loan size**, measured at the loan level by **log_loan_amount** (Section 5.2 only).
- **X: Economic development**, measured by **log_gdp_pc**.
- **Z: Institutional environment**, measured primarily by **institutional_pca1**, with alternative measures
  **institutional_index** and **financial_access_index**.
- **Controls**, especially **log_population**, to account for country size.
- **Alternative development proxy**, **poverty_rate**, used only as a limited-sample robustness exercise
  because missingness is substantial.

Before moving to figures and regressions, I audit the merged panel for duplicate country-year keys,
variable missingness, and within-country variation. This matters because the analysis depends on correct
ISO/year merges and because some merged variables may be available for fewer observations than the Kiva
outcome.


```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

panel_path = Path("outputs") / "final_panel_country_year.csv"
reg = pd.read_csv(panel_path)
print(f"Loaded panel from: {panel_path}")

numeric_cols = [
    "total_loan_amount",
    "log_total_loan_amount",
    "log_gdp_pc",
    "institutional_pca1",
    "institutional_index",
    "financial_access_index",
    "population",
    "log_population",
    "poverty_rate",
    "year"
]
for col in numeric_cols:
    if col in reg.columns:
        reg[col] = pd.to_numeric(reg[col], errors="coerce")

reg["country_code"] = reg["country_code"].astype(str).str.strip().str.upper()

print("Rows:", len(reg))
print("Columns:", reg.columns.tolist())

duplicate_keys = reg.duplicated(["country_code", "year"]).sum()
print("Duplicate country-year keys:", int(duplicate_keys))
print("Countries:", reg["country_code"].nunique(dropna=True))
print("Years:", reg["year"].nunique(dropna=True))

core_audit_cols = [
    "log_total_loan_amount", "log_gdp_pc", "institutional_pca1",
    "institutional_index", "financial_access_index", "log_population", "poverty_rate"
]
missing_share = reg[core_audit_cols].isna().mean().sort_values(ascending=False).round(3)
display(missing_share.to_frame("missing_share"))

variation_cols = ["institutional_pca1", "institutional_index", "financial_access_index", "population"]
within_country_nunique = reg.groupby("country_code")[variation_cols].nunique(dropna=True)
display(within_country_nunique.describe().round(2))

reg.head()
```

### 2.2 Merging with External Datasets

To enrich the country-year Kiva panel, I merge in multiple external datasets. These outside datasets are
essential because the original Kiva data describe lending activity, but they do not by themselves measure
cross-country development, institutional quality, or population size.

**First, I merge a World Bank GDP dataset** containing annual GDP per capita in constant 2015 USD. This
dataset is used to construct **gdp_pc_const2015** and **log_gdp_pc**, which serve as the main explanatory
variables for economic development.

**Second, I merge governance and financial-development indicators** from World Bank sources. These include
variables used to construct **institutional_index**, **institutional_pca1**, and **financial_access_index**.
These measures capture dimensions of institutional strength, governance quality, and access to formal
financial infrastructure, and they play the role of mechanism **Z**.

**Third, I merge country-level demographic and development controls**, especially **population**, and in
the broader cleaned dataset also measures such as **HDI**, **life expectancy**, and **GNI** where
available. Population is particularly important because raw loan totals partly reflect country size; using
it allows per-capita measures that avoid conflating scale with lending intensity.

All external datasets are organized at the **country-year level** and are merged using **country codes**
and **year**. The panel audit above reports duplicate country-year keys, missingness after merging, and
within-country variation in merged variables; some institutional or financial-access series are incomplete
or nearly time-invariant over 2013–2017, which is why `institutional_pca1`, `institutional_index`, and
`financial_access_index` are all reported and compared rather than relying on a single measure.


### 2.3 Sample Period Coverage Check (2013 vs. Later Years)

The panel begins in 2013, the year Kiva's own posting volume was still ramping up, so before treating the
2013–2017 year fixed effects in Section 5 as evidence of anything, it is worth checking directly whether
2013 is a genuinely comparable observation year or a partial/incomplete one that would bias the panel if
kept.

**Decision rule (applied mechanically, before looking at the regression results):** 2013 is treated as
*incomplete* and dropped from the main sample if either (a) fewer than 10 of 12 months are represented in
the raw loan-posting dates, or (b) loan counts are below 40% of 2014's *and* country coverage is clearly
lower than 2014's. Otherwise, 2013 is treated as a genuine (if smaller) year of real platform activity and
the full 2013–2017 sample is retained, with the burden shifted to explicitly interpreting the year effects
as reflecting Kiva's expansion rather than a data artifact.

Table A1 in the Appendix reports loan counts, total dollar volume, country coverage, and month-by-month
coverage for each year, built directly from the raw Kiva loan-posting timestamps (`posted_time`) rather
than from the aggregated panel, since date-level information does not survive aggregation to
`final_panel_country_year.csv`.


```python
year_coverage = pd.read_csv("outputs/table_year_month_coverage.csv")

months_missing_2013 = 12 - int(year_coverage.loc[year_coverage["year"] == 2013, "n_months_covered"].iloc[0])
pct_2013_of_2014 = float(year_coverage.loc[year_coverage["year"] == 2013, "pct_of_2014_n_loans"].iloc[0])
n_countries_2013 = int(year_coverage.loc[year_coverage["year"] == 2013, "n_countries"].iloc[0])
n_countries_2014 = int(year_coverage.loc[year_coverage["year"] == 2014, "n_countries"].iloc[0])

print("Full year-by-year detail: see Table A1 in the Appendix.")
print()
print(f"2013 months missing (of 12): {months_missing_2013}")
print(f"2013 loan count as % of 2014: {pct_2013_of_2014}%")
print(f"2013 countries: {n_countries_2013} vs. 2014: {n_countries_2014}")

incomplete_by_months = months_missing_2013 >= 3
incomplete_by_volume_and_countries = (pct_2013_of_2014 < 40) and (n_countries_2013 < 0.8 * n_countries_2014)
sample_period_decision = "DROP 2013 (incomplete year)" if (incomplete_by_months or incomplete_by_volume_and_countries) else "KEEP full 2013-2017 sample (real platform expansion)"
print()
print("Decision:", sample_period_decision)
```

Table A1 shows that 2013 has **complete 12-month coverage** with no gap of 3 or more months, and that
loan counts (140,034) are **80.4%** of 2014's, not below the 40% threshold. Country coverage (74) is also
close to 2014's (81), not "clearly lower" by the standard applied here. Neither disqualifying condition is
met, so **the decision is to keep the full 2013–2017 sample**: the lower activity levels in 2013 relative
to later years are treated as evidence of Kiva's real platform growth over this period rather than as a
partial-year data artifact. This directly motivates including year fixed effects in the preferred
specifications (m3–m4, m6–m8; lm3–lm4, lm6–lm8) rather than treating them as a nuisance control, and Section
5.1 returns to this point when interpreting the year coefficients. As an additional check given that this
judgment call could in principle be wrong, Section 6.2 re-estimates the preferred quadratic specifications
on the 2014–2017 subsample only and confirms the main conclusions are unchanged either way.


## 3. Descriptive Evidence

### 3.1 Summary Statistics

Table 1 reports summary statistics for the core regression variables. The sample spans substantial
variation in both development and lending intensity, which matters because the analysis below tests
whether the development–lending relationship is monotonic, nonlinear, or different across the quantity and
size margins.

Variation in **log_gdp_pc** indicates that the sample includes low-income, lower-middle income,
upper-middle income, and relatively richer economies. Variation in **log_total_loan_amount** shows that
Kiva lending is far from uniform across countries. The institutional and financial-access variables also
display meaningful dispersion, implying that governance and financial infrastructure differ markedly
across the sample.

Table 2 reports the same lending variable split by development tercile, which gives an early, purely
descriptive check of whether lending activity is monotonic or peaks somewhere in the middle of the
development distribution.


```python
core_summary_vars = [
    "log_total_loan_amount",
    "log_gdp_pc",
    "institutional_pca1",
    "institutional_index",
    "financial_access_index",
    "log_population",
    "poverty_rate"
]

summary_table = reg[core_summary_vars].agg([
    "count", "mean", "std", "min", "median", "max"
]).T
summary_table = summary_table.rename(columns={
    "count": "N",
    "mean": "Mean",
    "std": "SD",
    "min": "Min",
    "median": "Median",
    "max": "Max"
}).round(3)
print("Table 1. Summary Statistics")
display(summary_table)
```

```python
tercile_cols = [
    "log_total_loan_amount",
    "log_gdp_pc",
    "institutional_pca1",
    "log_population"
]

summary_df = reg[tercile_cols].dropna().copy()
summary_df["development_tercile"] = pd.qcut(
    summary_df["log_gdp_pc"],
    q=3,
    labels=["Low development", "Middle development", "High development"],
    duplicates="drop"
)

subgroup_table = summary_df.groupby("development_tercile", observed=False).agg(
    observations=("log_total_loan_amount", "size"),
    mean_log_loan=("log_total_loan_amount", "mean"),
    mean_log_gdp_pc=("log_gdp_pc", "mean"),
    mean_institution=("institutional_pca1", "mean"),
    mean_log_population=("log_population", "mean")
).round(3)

print("Table 2. Lending Activity by Development Tercile")
display(subgroup_table)
```

The middle-development tercile has the highest mean of `log_total_loan_amount` (13.44) compared to the
low- (12.81) and high-development terciles (12.78). This raw, bivariate pattern is the origin of the
original middle-income-peak hypothesis, and it is worth taking seriously as a starting point. However, as
Section 5.1 shows, this pattern does not survive as a statistically significant nonlinear relationship once
the regressions control for institutions, population, and year effects and formally test the quadratic
term. The tercile comparison should therefore be read as a motivating pattern rather than as confirmatory
evidence on its own.


### 3.2 Figures

The figures below serve several purposes: to visualize the core bivariate relationship, to show the
distribution of the dependent variable, to show the distribution of the main explanatory variable, to
rescale lending by population, and to check whether the same patterns are visible geographically.

**Figure 1** plots the main bivariate relationship between development and total lending volume, with a
quadratic fit and a LOWESS smoother.
**Figure 2** shows the distribution of lending across country-years.
**Figure 3** shows the distribution of log GDP per capita across the analysis sample.
**Figure 4** rescales lending by population and asks whether the development pattern looks similar in
per-capita terms.
**Figures 5–7** show the same variables mapped across countries.


```python
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess

plot_df = reg[["log_gdp_pc", "log_total_loan_amount"]].replace([np.inf, -np.inf], np.nan).dropna()

x = plot_df["log_gdp_pc"].to_numpy()
y = plot_df["log_total_loan_amount"].to_numpy()

coef = np.polyfit(x, y, 2)
poly = np.poly1d(coef)
x_grid = np.linspace(x.min(), x.max(), 200)
y_grid = poly(x_grid)
low = lowess(y, x, frac=0.4, return_sorted=True)
descriptive_turning_point = -coef[1] / (2 * coef[0]) if coef[0] != 0 else np.nan

plt.figure(figsize=(7, 4.5), dpi=150)
plt.scatter(x, y, alpha=0.35)
plt.plot(x_grid, y_grid, linewidth=2, label="Quadratic fit")
plt.plot(low[:, 0], low[:, 1], linewidth=2, label="LOWESS")
if np.isfinite(descriptive_turning_point) and x.min() <= descriptive_turning_point <= x.max():
    plt.axvline(descriptive_turning_point, linestyle="--", linewidth=1, color="grey")
plt.xlabel("Log GDP per capita")
plt.ylabel("Log total loan amount")
plt.title("Figure 1. Development and Total Lending Volume")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

**Figure 1.** The scatterplot with both a fitted quadratic curve and a LOWESS smoother is the descriptive
origin of the middle-income-peak hypothesis: both curves bend downward at higher income levels. This is
still purely descriptive, unconditional evidence — it does not control for institutions, population, or
year, and Section 5.1 shows the same nonlinearity is not statistically significant once those controls and
a formal test are added.


```python
plt.figure()
plt.hist(reg["log_total_loan_amount"].dropna(), bins=20)
plt.xlabel("Log Total Loan Amount")
plt.ylabel("Frequency")
plt.title("Figure 2. Distribution of Kiva Lending Across Country-Years")
plt.show()
```

**Figure 2.** The distribution of the dependent variable shows substantial heterogeneity in Kiva lending
across country-years. This matters because the analysis relies on meaningful cross-country variation
rather than only a few narrowly clustered observations.


```python
plt.figure()
plt.hist(reg["log_gdp_pc"].dropna(), bins=20)
plt.xlabel("Log GDP per Capita")
plt.ylabel("Frequency")
plt.title("Figure 3. Distribution of Economic Development in the Sample")
plt.show()
```

**Figure 3.** The histogram of log GDP per capita confirms that the sample covers a wide development range.
That breadth is necessary for evaluating whether the development–microfinance relationship is monotonic,
nonlinear, or different across the quantity and size margins.


```python
plot_df4 = reg[["log_gdp_pc", "total_loan_amount", "population"]].replace([np.inf, -np.inf], np.nan).dropna().copy()
plot_df4 = plot_df4[plot_df4["population"] > 0].copy()

plot_df4["loan_per_100k"] = plot_df4["total_loan_amount"] / plot_df4["population"] * 100000
# loan_per_100k is strictly positive whenever total_loan_amount > 0 (true for every row in this
# panel, since it is built from realized Kiva loans), so a plain log is used throughout the paper
# rather than log1p -- see Section 2 for the single per-capita transform used in both this figure
# and the maps in Section 3.3.
plot_df4["log_loan_per_100k"] = np.log(plot_df4["loan_per_100k"])

x4 = plot_df4["log_gdp_pc"].to_numpy()
y4 = plot_df4["log_loan_per_100k"].to_numpy()

coef4 = np.polyfit(x4, y4, 2)
poly4 = np.poly1d(coef4)
x4_grid = np.linspace(x4.min(), x4.max(), 200)

plt.figure(figsize=(7, 4.5), dpi=150)
plt.scatter(x4, y4, alpha=0.35)
plt.plot(x4_grid, poly4(x4_grid), linewidth=2)
plt.xlabel("Log GDP per capita")
plt.ylabel("Log loan amount per 100k people")
plt.title("Figure 4. Per-Capita Microfinance Lending and Economic Development")
plt.tight_layout()
plt.show()
```

**Figure 4.** This figure rescales lending by population, using loan amount per 100,000 people rather than
raw loan totals, which reduces the risk of mechanically classifying very large countries as high-lending
simply because they are large. The per-capita pattern is broadly similar in shape to Figure 1.


### 3.3 Maps

The next three figures use static cross-country maps to show how lending activity, development, and
institutions are distributed spatially. The goal is not causal identification, but to check whether the
geography of microfinance is visually consistent with the patterns in Section 3.2. Each map averages
country-year values over 2013–2017 to produce one static country value, and merge diagnostics are printed
so that unmatched ISO codes are visible rather than hidden.

**Figure 5** maps the main outcome in per-capita terms.
**Figure 6** maps the main X variable, economic development.
**Figure 7** maps the institutional mechanism Z.

The map merge uses the shapefile's `ISO_A2_EH` field rather than `ISO_A2`. In the Natural Earth 110m
dataset, `ISO_A2` is coded `"-99"` for five territories with complex sovereignty status (France, Norway,
Northern Cyprus, Somaliland, and Kosovo). The original code filtered `world` on `ISO_A2 != "-99"` before
merging, which dropped all five countries' geometry from the map entirely — not just their data, but their
outline — leaving holes in the choropleth rather than a "no data" grey fill. Of these five, only **Kosovo
(XK)** is a Kiva-active country in this panel, so the fix directly recovers one real data point that was
previously invisible on the map; the other four (including France and Norway) are not in the panel at all
and were never going to show lending data, but the fix still restores their outlines so the map reads as
geographically complete rather than having unexplained gaps. `ISO_A2_EH` carries the correct two-letter
code for all five cases and is used throughout, with no `-99` filter applied.


```python
import geopandas as gpd
from pathlib import Path

map_df = reg.copy()
map_df = map_df[map_df["population"].notna() & (map_df["population"] > 0)].copy()
map_df["loan_per_100k"] = map_df["total_loan_amount"] / map_df["population"] * 100000
map_df["log_loan_per_100k"] = np.log(map_df["loan_per_100k"])

map_df = map_df.groupby("country_code", as_index=False).mean(numeric_only=True)
map_df["country_code"] = map_df["country_code"].astype(str).str.strip().str.upper()

shp_path = Path("data/natural_earth_countries_110m/ne_110m_admin_0_countries.shp")
world = gpd.read_file(shp_path)
world["ISO_A2_EH"] = world["ISO_A2_EH"].astype(str).str.strip().str.upper()

unmatched_data_codes = sorted(set(map_df["country_code"]) - set(world["ISO_A2_EH"]))
print("Unmatched country codes in panel:", unmatched_data_codes)

world = world.merge(map_df, left_on="ISO_A2_EH", right_on="country_code", how="left")
print("Matched shapefile countries:", int(world["country_code"].notna().sum()))
print(
    "Kosovo (the only -99-coded country that is Kiva-active in this panel) has data:",
    bool(world.loc[world["ISO_A2_EH"] == "XK", "country_code"].notna().any())
)

fig, ax = plt.subplots(figsize=(12, 6), dpi=150)
try:
    world.plot(
        column="log_loan_per_100k",
        cmap="viridis",
        legend=True,
        scheme="quantiles",
        k=5,
        ax=ax,
        missing_kwds={"color": "lightgrey"}
    )
except Exception:
    world.plot(
        column="log_loan_per_100k",
        cmap="viridis",
        legend=True,
        ax=ax,
        missing_kwds={"color": "lightgrey"}
    )

ax.set_title("Figure 5. Microfinance Lending per 100k People (2013–2017 Average)")
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

**Figure 5.** Average microfinance lending per 100,000 people across countries. Using a per-capita measure
is important because raw loan totals partly reflect country size rather than lending intensity. With the
`ISO_A2_EH` fix, Kosovo now displays its actual lending data instead of being an invisible gap in the map,
and France, Norway, Northern Cyprus, and Somaliland are restored as grey "no data" territories rather than
holes — this panel has no Kiva-lending observations for the latter four, so grey is the correct rendering
for them, not colored intensity.


```python
fig, ax = plt.subplots(figsize=(12, 6), dpi=150)
try:
    world.plot(
        column="log_gdp_pc",
        cmap="viridis",
        legend=True,
        scheme="quantiles",
        k=5,
        ax=ax,
        missing_kwds={"color": "lightgrey"}
    )
except Exception:
    world.plot(
        column="log_gdp_pc",
        cmap="viridis",
        legend=True,
        ax=ax,
        missing_kwds={"color": "lightgrey"}
    )
ax.set_title("Figure 6. Economic Development: Log GDP per Capita (2013–2017 Average)")
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

## 4. Empirical Strategy

The regression section tests whether the descriptive patterns in Section 3 survive formal specifications,
at both the country-year and loan level. I estimate two families of models, one per outcome:

- **Quantity models** (Section 5.1): dependent variable `log_total_loan_amount`, country-year panel,
  country-clustered standard errors.
- **Size models** (Section 5.2): dependent variable `log_loan_amount`, loan-level panel,
  country-year-clustered standard errors (with a country-clustered robustness check in Section 6).

For each outcome I report four linear/interaction specifications and four quadratic/robustness
specifications:

1. **Linear benchmark** — development enters monotonically.
2. **Linear + institutions** — adds the institutional-quality control.
3. **+ Year fixed effects** — absorbs common shocks over time.
4. **+ Interaction** — tests whether the development slope varies with institutional quality.
5–8. **Quadratic and robustness** — adds a squared development term (centered, to ease interpretation of
   the linear term at the sample mean), then swaps in year fixed effects and alternative institutional
   measures (`institutional_index`, `financial_access_index`).

Both outcomes are tested with the same battery so that any divergence between them is a difference in the
underlying relationship, not a difference in specification. Because a squared term can fit noise as
readily as a genuine peak, the analysis treats **statistical significance of the quadratic term**, not just
its sign, as the criterion for whether a nonlinear pattern is supported — for either outcome.


```python
import statsmodels.formula.api as smf
from scipy import stats

def fit_ols(formula, data, se=None, cluster_col=None):
    if se == "cluster" and cluster_col is None:
        raise ValueError("cluster_col must be provided when se='cluster'.")

    model = smf.ols(formula=formula, data=data, missing="drop")
    used_rows = model.data.row_labels

    if se == "cluster":
        groups = data.loc[used_rows, cluster_col]
        result = model.fit(cov_type="cluster", cov_kwds={"groups": groups})
    else:
        result = model.fit()

    return result


def sample_audit(data, cols):
    cols = [c for c in cols if c in data.columns]
    used = data[cols].dropna().copy()
    return pd.Series({
        "obs": len(used),
        "countries": used["country_code"].nunique() if "country_code" in used.columns else np.nan,
        "years": used["year"].nunique() if "year" in used.columns else np.nan,
        "mean_log_gdp_pc": used["log_gdp_pc"].mean() if "log_gdp_pc" in used.columns else np.nan
    })


def interaction_margins(model, x_var, z_var, z_values):
    """Marginal effect of x_var on the outcome at given values of z_var, via the delta method,
    for a model that includes x_var, z_var, and their product term x_var:z_var."""
    inter = f"{x_var}:{z_var}"
    b_x, b_xz = model.params[x_var], model.params[inter]
    cov = model.cov_params()
    var_x, var_xz, cov_x_xz = cov.loc[x_var, x_var], cov.loc[inter, inter], cov.loc[x_var, inter]

    rows = []
    for label, z in z_values.items():
        me = b_x + b_xz * z
        se = np.sqrt(var_x + (z ** 2) * var_xz + 2 * z * cov_x_xz)
        t = me / se
        p = 2 * (1 - stats.norm.cdf(abs(t)))
        rows.append({"institutional_pca1 level": label, "value": round(z, 3),
                     "marginal_effect": round(me, 4), "se": round(se, 4), "p_value": round(p, 4)})
    return pd.DataFrame(rows)
```

Model **m4** adds an interaction between development and institutional quality. Table 3b reports the

Table 4 is the direct test of the middle-income-peak hypothesis at the country level, and it does **not**
support it. The squared term (`c_log_gdp_pc_sq`) is negative in all four specifications, ranging from
-0.155 to -0.201, but is **not statistically distinguishable from zero in any of them** (p ranges from
about 0.10 to 0.18, with standard errors around 0.12). The point estimates imply a concave curve with a
turning point inside the sample range (Table 4's diagnostic cell above), but because the quadratic
coefficient itself is not significant, the data cannot reject a much simpler description — a flat or
monotonically declining relationship — over the "peak" shape implied by the point estimate.

Consistent with that reading, the linear component evaluated at the sample mean (`c_log_gdp_pc`) is also
negative in every specification (m5–m8), and total lending volume's estimated relationship with
development is a decline across most of the observed income range, not a rise to a peak. Models m6–m8 show
that this pattern is stable whether institutional quality is measured with `institutional_pca1`,
`institutional_index`, or `financial_access_index`, and whether or not year fixed effects are included —
the instability is in the *magnitude and significance* of the curvature, not in which control variables are
used.

**We do not treat the country-level quadratic pattern as evidence of a middle-income peak in total lending
volume.** The more defensible summary of Table 4 is that total Kiva lending volume tends to be
lower, not higher, in richer countries within this sample, without strong evidence of a rise-then-fall
shape.


**A note on the year fixed effects.** Section 2.3 established that 2013 is a complete, comparable
observation year rather than a partial one, so the year coefficients in m3, m4, and m6–m8 are not
correcting for a data artifact. In model m6, the year dummies rise from 0 (2013, the omitted baseline) to
roughly **+2.3 to +2.9 log points** in 2014–2017, with no further monotonic increase after 2014 — consistent
with Kiva's total lending volume genuinely expanding over this period (Table A1: total dollar volume rises
from about \$133.0M in 2013 to \$175.6M in 2017) rather than with a sample-composition change, since country
coverage does not grow nearly as much (74 to 76 countries) as dollar volume does. This supports reading the
year effects as a platform-growth control rather than as evidence bearing on the development–lending
relationship itself.

individual loans rather than 348 country-years, so these models have much more power to detect a nonlinear
pattern if one exists. Table 5b reports the marginal effect of development on loan size at low, average,
and high institutional quality for the interaction model (lm4), again via the delta method.

Table 6's squared term (`c_log_gdp_pc_sq`) is **positive and statistically significant at the 1% level**
(approximately +0.245 in the preferred specification), in every quadratic specification. This is a
**U-shape, not an inverted U**: individual loan size is estimated to fall as development rises from the
low end of the sample, reach a minimum, and then rise again at higher development levels. This sign is the
**opposite** of the (statistically insignificant) country-level pattern in Table 4.

This is not a contradiction, because Tables 4 and 6 measure different objects. Table 4 asks whether a
country's *total* Kiva lending volume varies with development; Table 6 asks whether the *typical size of a
loan*, conditional on a loan being made, varies with development. A country can plausibly receive less
total lending as it develops (fewer, larger loans crowd out many small ones, and formal banking substitutes
for Kiva overall) while the loans that *are* still made through Kiva get smaller in lower-middle income
settings — where Kiva serves a high volume of very small borrowers — before rising again as remaining Kiva
borrowers in richer countries use it for larger, more bank-like loan sizes. Sections 5.1 and 5.2 should
therefore be read as two separate, both well-supported findings about different margins of microfinance
activity, not as two tests of one shared hypothesis.


### 5.3 Two Margins: Extensive vs. Intensive

Sections 5.1 and 5.2 both condition on Kiva already lending in a country-year — as Section 7 (Limitations)
notes, countries with no Kiva presence simply do not appear in `final_panel_country_year.csv` at all. That
means Table 4's insignificant quadratic term cannot distinguish two very different stories: (i) development
has no nonlinear relationship with Kiva activity at any margin, or (ii) development *does* have a nonlinear
relationship with **whether Kiva is active in a country at all** (the extensive margin), but that pattern
is invisible in Table 4 because the regression only ever sees countries where the answer is already "yes."

To test this, I build a **global country-year panel that includes true zeros**: every country with World
Bank GDP-per-capita (`NY.GDP.PCAP.KD`, constant 2015 USD) and population (`SP.POP.TOTL`) data for
2013–2017, outer-merged with Kiva country-year loan totals built directly from the raw loan-posting data.
Country-years with no matching Kiva loans get `total_loan_amount = 0`, `n_loans = 0`, and a `kiva_active`
indicator of 0. Country codes are harmonized to ISO3 using `country_converter` (Kiva's own two-letter
codes were converted via `ISO2 → ISO3`, with all 90 Kiva country codes — including territory-style codes
such as `XK` for Kosovo and `PS` for the West Bank and Gaza — resolving cleanly). WDI classifies 217
economies as actual countries (excluding regional/income aggregates such as "Sub-Saharan Africa"); of the
resulting 1,085 country-year cells, 65 are dropped for missing GDP per capita or population (mostly very
small territories and a few years of state fragility), leaving **1,020 country-years covering 206
countries**, of which **374 (36.7%) are Kiva-active**.

Because institutional and financial-access variables are only available for the ~90–96 countries Kiva
already covers, they are heavily missing in the global sample and are **not used as controls in the global
regressions** below — the global specification is deliberately parsimonious (development, population, year
fixed effects). `institutional_pca1` is reintroduced only in the Kiva-only PPML column, where coverage is
adequate, consistent with the rest of the paper.

I estimate two models on this global panel:

- **Extensive margin — Logit.** `kiva_active ~ c_log_gdp_pc + c_log_gdp_pc_sq + log_population + C(year)`,
  country-clustered SE, reporting both raw coefficients and average marginal effects (AME).
- **Intensive margin — PPML (Poisson pseudo-maximum-likelihood).** Because the outcome now includes true
  zeros, a log-linear OLS model is not an option (`log(0)` is undefined and dropping zeros just reproduces
  Section 5.1's selected sample). PPML handles zeros natively and is estimated on (i) the full global
  sample including zeros and (ii) the Kiva-only subsample, so the same functional form can be compared with
  and without the zero-lending country-years.


```python
import statsmodels.api as sm

global_path = Path("outputs/final_panel_global_extensive.csv")
gpanel = pd.read_csv(global_path)
print("Loaded global extensive-margin panel from:", global_path)
print("Shape:", gpanel.shape)
print("Countries:", gpanel["iso3"].nunique())
print("Kiva-active country-years:", int(gpanel["kiva_active"].sum()), f"({gpanel['kiva_active'].mean():.1%})")
gpanel.head()
```

```python
# Extensive margin: Logit(kiva_active ~ development), country-clustered SE
logit_formula = "kiva_active ~ c_log_gdp_pc + c_log_gdp_pc_sq + log_population + C(year)"
logit_model = smf.logit(logit_formula, data=gpanel, missing="drop")
logit_rows = logit_model.data.row_labels
logit_groups = gpanel.loc[logit_rows, "iso3"]
logit_fit = logit_model.fit(cov_type="cluster", cov_kwds={"groups": logit_groups}, disp=0)

logit_margeff = logit_fit.get_margeff()

print("Table 8a. Extensive Margin: Logit Coefficients (Kiva Active in Country-Year)")
display(logit_fit.summary2().tables[1].round(4))

print("Table 8b. Extensive Margin: Average Marginal Effects (AME)")
display(logit_margeff.summary_frame().round(4))

logit_b1 = logit_fit.params["c_log_gdp_pc"]
logit_b2 = logit_fit.params["c_log_gdp_pc_sq"]
logit_mean_log_gdp = gpanel["log_gdp_pc"].mean()
if logit_b2 < 0:
    logit_tp = logit_mean_log_gdp - logit_b1 / (2 * logit_b2)
    print()
    print("Logit turning point (log GDP per capita):", round(logit_tp, 3), "| level (const. 2015 USD):", round(np.exp(logit_tp), 2))
    print("Sample range:", round(gpanel['log_gdp_pc'].min(), 3), "to", round(gpanel['log_gdp_pc'].max(), 3))
```

```python
# Intensive margin: PPML on (i) the global sample including zeros, (ii) the Kiva-only subsample.
# total_loan_amount is rescaled to thousands of USD for numerical stability in the Poisson IRLS fit.
gpanel["total_loan_amount_k"] = gpanel["total_loan_amount"] / 1000.0

ppml_global_formula = "total_loan_amount_k ~ c_log_gdp_pc + c_log_gdp_pc_sq + log_population + C(year)"
ppml_global_model = smf.glm(ppml_global_formula, data=gpanel, family=sm.families.Poisson(), missing="drop")
ppml_global_rows = ppml_global_model.data.row_labels
ppml_global_groups = gpanel.loc[ppml_global_rows, "iso3"]
ppml_global_fit = ppml_global_model.fit(cov_type="cluster", cov_kwds={"groups": ppml_global_groups})

kiva_only = gpanel[gpanel["kiva_active"] == 1].copy()
ppml_kiva_formula = "total_loan_amount_k ~ c_log_gdp_pc + c_log_gdp_pc_sq + institutional_pca1 + log_population + C(year)"
ppml_kiva_model = smf.glm(ppml_kiva_formula, data=kiva_only, family=sm.families.Poisson(), missing="drop")
ppml_kiva_rows = ppml_kiva_model.data.row_labels
ppml_kiva_groups = kiva_only.loc[ppml_kiva_rows, "iso3"]
ppml_kiva_fit = ppml_kiva_model.fit(cov_type="cluster", cov_kwds={"groups": ppml_kiva_groups})

print("PPML converged (global, incl. zeros):", ppml_global_fit.converged, "| N =", int(ppml_global_fit.nobs))
print("PPML converged (Kiva-only):", ppml_kiva_fit.converged, "| N =", int(ppml_kiva_fit.nobs))

logit_ame = logit_margeff.summary_frame()["dy/dx"]

print()
print("Table 9. Two Margins: Logit (Extensive) vs. PPML (Intensive), Development Terms")
key_rows = ["c_log_gdp_pc", "c_log_gdp_pc_sq"]
table9 = pd.DataFrame({
    "Logit coef.": logit_fit.params.loc[key_rows],
    "Logit p": logit_fit.pvalues.loc[key_rows],
    "Logit AME": logit_ame.loc[key_rows],
    "PPML global coef.": ppml_global_fit.params.loc[key_rows],
    "PPML global p": ppml_global_fit.pvalues.loc[key_rows],
    "PPML Kiva-only coef.": ppml_kiva_fit.params.loc[key_rows],
    "PPML Kiva-only p": ppml_kiva_fit.pvalues.loc[key_rows],
}).round(4)
display(table9)
```

Table 9 shows where the strongest evidence for development-related nonlinearity lives — with an
important qualification added by the formal test in Section 5.4. The **extensive margin is
significantly concave**: the Logit quadratic term is negative and significant (coefficient −0.260, p
= 0.023; AME −0.040, p = 0.018), and Figure 8 shows the pattern directly in the raw data — Kiva's
entry rate is roughly two-thirds across the poorest three deciles (peaking at 72% in the second
decile), falls to half by the fourth decile, and declines steeply to between 1% and 7% in the top
three deciles. The implied turning point is at log GDP per capita ≈ 6.79 (about \$893 in constant
2015 USD), very close to the low end of the sample. Applying the same Lind–Mehlum test used for the
other margins (Section 5.4, Table 10), the high-end decline is unambiguous but the low-end *rise* is
not individually significant (t = 1.07, one-sided p = 0.143), so the data confirm concavity and a
steep decline with development — not a completed rise-then-fall within the observed range.

The **intensive margin does not show this pattern, whether or not zeros are included.** PPML on the
full global sample (including the 646 zero-lending country-years) gives a quadratic coefficient of
−0.232 (p = 0.242); PPML restricted to the Kiva-active country-years (369 of the 374; five drop for
missing institutional controls) gives −0.070 (p = 0.530). Both are statistically indistinguishable
from zero, echoing the OLS-in-logs result in Table 4 (quadratic term −0.155 to −0.201, also
insignificant) even though PPML uses a completely different estimator and, in the global
specification, a completely different (zero-inclusive) sample.

This resolves the ambiguity raised at the start of Section 5.3 — with more nuance than a simple "the
inverted-U was hiding at the extensive margin." Development does matter most visibly for **whether**
Kiva operates in a country: entry probability is significantly concave in income and falls steeply
as countries develop. But by the paper's own formal standard, that is not the same as a confirmed
inverted-U, because the initial rise at the very bottom of the income distribution cannot be
statistically distinguished from a flat start. The three-margin picture is therefore: **(1) whether
Kiva enters a country — significantly concave, steeply declining, peak unconfirmed; (2) how much a
country receives in total, conditional on entry — flat/declining, not significant; (3) how large an
individual loan is — U-shaped, significant and formally confirmed** (Section 5.2). These are three
distinct empirical objects, and the original "middle-income peak" intuition survives, at most, in
weakened form at the first of them.

### 5.4 Formal Test for Nonlinearity: The Lind–Mehlum (2010) U-Test

Throughout Sections 5.1–5.3, "the quadratic term is/isn't significant" has been the criterion for whether a
nonlinear pattern is taken seriously. That criterion is necessary but not sufficient: a significant
quadratic coefficient does not by itself guarantee that the curve actually turns within the observed data
range, because a significant coefficient combined with a turning point far outside the sample would imply
an essentially monotonic relationship over the range that is actually observed. Lind and Mehlum (2010)
formalize the correct test: a genuine U-shape (or inverted-U) requires that the slope of the curve be
**significantly negative (positive) at the low end of the sample and significantly positive (negative) at
the high end**, evaluated at the sample's actual minimum and maximum of X, not just that the quadratic
coefficient differs from zero at the sample mean.

Concretely, for a fitted quadratic $\hat y = \hat\beta_0 + \hat\beta_1 x + \hat\beta_2 x^2 + \text{controls}$,
the slope at any point is $\hat\beta_1 + 2\hat\beta_2 x$. I compute this slope at $x_{\min}$ and $x_{\max}$
(the sample extremes of centered log GDP per capita), obtain its standard error via the delta method
($\mathrm{Var}(\hat\beta_1) + (2x)^2\mathrm{Var}(\hat\beta_2) + 2(2x)\mathrm{Cov}(\hat\beta_1,\hat\beta_2)$),
and test both slopes individually. Following Lind and Mehlum's intersection-union logic (closely related to
Sasabuchi, 1980), the overall test statistic is the **weaker of the two individual $t$-statistics** (in
absolute value) — the shape is only confirmed if *both* ends pass, so the binding constraint is whichever
end is less significant. I also report a **Fieller (1954) confidence interval** for the turning point
$x_0=-\hat\beta_1/(2\hat\beta_2)$, which correctly accounts for the fact that $x_0$ is a ratio of two
estimated, correlated coefficients (a naive delta-method CI for a ratio can behave badly and is not used
here); if the quadratic term is not precisely estimated relative to the critical value, the Fieller interval
is unbounded, which is itself informative.

I apply this test to all three shape claims in the paper: **m6** (country-year, total lending
volume), **lm6** (loan-level, individual loan size), and the **extensive-margin Logit** from Section
5.3, the last evaluated on the linear-index (log-odds) scale where the quadratic functional form
lives.

```python
def lind_mehlum_utest(model, b1_name, b2_name, x_min, x_max, alpha=0.05):
    """Lind & Mehlum (2010) U-test for a quadratic term in a fitted statsmodels result.

    model: fitted results object containing coefficients b1_name (linear) and b2_name (quadratic)
    x_min, x_max: sample extremes of the (centered) X variable, on the same scale as b1_name/b2_name
    Returns a dict with the slope/t-stat at each end, the intersection-union test statistic and its
    one-sided p-value, whether the shape is confirmed, and a Fieller (1954) CI for the turning point.
    """
    b1, b2 = model.params[b1_name], model.params[b2_name]
    cov = model.cov_params()
    var_b1, var_b2 = cov.loc[b1_name, b1_name], cov.loc[b2_name, b2_name]
    cov_b1b2 = cov.loc[b1_name, b2_name]

    slope_lo = b1 + 2 * b2 * x_min
    slope_hi = b1 + 2 * b2 * x_max
    se_lo = np.sqrt(var_b1 + (2 * x_min) ** 2 * var_b2 + 2 * (2 * x_min) * cov_b1b2)
    se_hi = np.sqrt(var_b1 + (2 * x_max) ** 2 * var_b2 + 2 * (2 * x_max) * cov_b1b2)
    t_lo, t_hi = slope_lo / se_lo, slope_hi / se_hi

    opposite_signs = (slope_lo * slope_hi) < 0
    u_test_t = min(abs(t_lo), abs(t_hi))
    u_test_p = 1 - stats.norm.cdf(u_test_t)
    shape_confirmed = bool(opposite_signs and (u_test_p < alpha))

    z = stats.norm.ppf(1 - alpha / 2)
    A = b2 ** 2 - z ** 2 * var_b2
    B = b1 * b2 - z ** 2 * cov_b1b2
    C = b1 ** 2 - z ** 2 * var_b1
    disc = B ** 2 - A * C
    x0 = -b1 / (2 * b2)
    if A > 0 and disc >= 0:
        r1, r2 = (B - np.sqrt(disc)) / A, (B + np.sqrt(disc)) / A
        x0_ci = tuple(sorted([-r1 / 2, -r2 / 2]))
        fieller_valid = True
    else:
        x0_ci = (-np.inf, np.inf)
        fieller_valid = False

    return {
        "slope_low": slope_lo, "t_low": t_lo,
        "slope_high": slope_hi, "t_high": t_hi,
        "opposite_signs": opposite_signs,
        "u_test_t": u_test_t, "u_test_p": u_test_p,
        "shape_confirmed": shape_confirmed,
        "turning_point_centered": x0, "turning_point_ci_centered": x0_ci, "fieller_valid": fieller_valid,
    }


utest_m6 = lind_mehlum_utest(m6, "c_log_gdp_pc", "c_log_gdp_pc_sq", reg["c_log_gdp_pc"].min(), reg["c_log_gdp_pc"].max())
utest_lm6 = lind_mehlum_utest(lm6, "c_log_gdp_pc", "c_log_gdp_pc_sq", loan_reg["c_log_gdp_pc"].min(), loan_reg["c_log_gdp_pc"].max())

# Extensive-margin Logit: same test, applied on the linear-index (log-odds) scale, where the
# quadratic functional form actually lives. cov_params() is the fitted model's full country-clustered
# covariance matrix, so Cov(b1, b2) is not assumed to be zero.
utest_logit = lind_mehlum_utest(
    logit_fit, "c_log_gdp_pc", "c_log_gdp_pc_sq",
    gpanel["c_log_gdp_pc"].min(), gpanel["c_log_gdp_pc"].max(),
)

utest_table = pd.DataFrame({
    "m6 (country-year, volume)": utest_m6,
    "lm6 (loan-level, size)": utest_lm6,
    "Logit (global, extensive margin)": utest_logit,
}).T
utest_table["turning_point_ci_centered"] = utest_table["turning_point_ci_centered"].apply(
    lambda ci: f"[{ci[0]:.3f}, {ci[1]:.3f}]" if np.isfinite(ci[0]) else "unbounded (quadratic term not precisely enough estimated)"
)
print("Table 10. Lind-Mehlum (2010) U-Test: m6, lm6, and the Extensive-Margin Logit")
display(utest_table.round(4))

print()
print("Extensive-margin Logit U-test, key figures for in-text reporting:")
print(f"  slope_low  = {utest_logit['slope_low']:.4f}   t_low  = {utest_logit['t_low']:.3f}   "
      f"one-sided p_low  = {1 - stats.norm.cdf(utest_logit['t_low']):.4f}")
print(f"  slope_high = {utest_logit['slope_high']:.4f}   t_high = {utest_logit['t_high']:.3f}   "
      f"one-sided p_high = {1 - stats.norm.cdf(utest_logit['t_high']):.4f}")
print(f"  u_test_t = {utest_logit['u_test_t']:.3f}   u_test_p = {utest_logit['u_test_p']:.4f}   "
      f"shape_confirmed = {utest_logit['shape_confirmed']}")
tp_ci = utest_table.loc["Logit (global, extensive margin)", "turning_point_ci_centered"]
print(f"  Fieller 95% CI for turning point (centered log GDP pc): {tp_ci}")
```

Table 10 confirms, formally, what Section 5.1's p-value-only discussion already suggested informally. For
**m6**, the slope at the low end of the sample is positive but only $t \approx 1.08$, well short of
conventional significance, so the weaker-end test statistic (which is what the intersection-union logic
requires to pass) fails decisively (one-sided $p \approx 0.14$), and the Fieller CI for the turning point is
**unbounded** — the data are consistent with a turning point essentially anywhere, including outside the
observed range entirely. This is additional, independent confirmation (beyond Table 4's own insignificant
quadratic coefficient) that **the country-level "peak" should not be treated as an established finding**:
it fails even a test designed to be maximally charitable to a curved relationship, since it only asks that
*both ends* individually clear a significance bar, and one end does not come close.

For **lm6**, the result is the opposite in every respect: the slope is significantly negative at the low
end ($t \approx -5.75$) and significantly positive at the high end ($t \approx 10.95$), both comfortably
clearing the bar, so the U-test **confirms** the shape (one-sided $p < 0.001$). The Fieller CI for the
turning point is tight and lies well inside the sample range, at roughly log GDP per capita
$\approx 7.05$–$7.59$ (about \$1,150–\$1,970 in constant 2015 USD), bracketing the point estimate of
$\approx 7.36$ (\$1,577) — a materially informative interval, not an artifact of a mechanically
significant-looking quadratic term. **This is the formal, quantitative basis for why the paper treats the
loan-level U-shape as the paper's central nonlinear finding and the country-level curvature as
unsupported**, rather than presenting both as symmetric "peak" and "trough" results of equal standing.

For the **extensive-margin Logit**, the test delivers a split verdict that Section 5.3's headline
should — and now does — reflect. The high-end slope is strongly negative and individually
significant (t = -3.03), confirming the steep decline in entry probability among richer countries.
But the low-end slope, while positive, is not individually significant (t = 1.07, one-sided p =
0.143), so the intersection-union test fails and the inverted-U is not formally confirmed: the data
cannot rule out that entry probability is simply flat at the very bottom of the income distribution
before declining. This is the same discipline that Section 5.1's country-level curvature was held
to, applied symmetrically to the paper's most favorable-looking result — the extensive margin is
reported as significant concavity with a steep decline, not as an established peak.

### 6.3 Loan-Level Robustness: Alternative Specifications of the Size Margin

The loan-level U-shape in Table 6 is the paper's central nonlinear finding, so it is worth stress-testing
directly rather than only checking the clustering choice (Section 6.1) and the sample period (Section 6.2).
I re-estimate lm6 under three additional variants:

1. **Excluding the United States.** US-based Kiva loans are a qualitatively different product (small
   business loans in a high-income economy) rather than the poverty-lending use case the rest of the panel
   represents, and the US contributes a large share of upper-income observations, so it could be driving
   the "rises again" half of the U-shape.
2. **Relative loan size**, `log(loan_amount / gdp_pc_const2015)`, instead of the nominal-dollar
   `log_loan_amount`. This targets a specific concern: if richer countries simply have higher price levels
   and costs of living, a nominal-dollar U-shape could partly reflect currency/price-level scaling rather
   than a real change in typical loan size relative to the local economy.
3. **Sector fixed effects.** If richer and poorer countries in the sample differ systematically in which
   Kiva sectors are active (e.g., more agriculture loans in low-income countries, more retail or services
   in higher-income ones), the loan-size U-shape could partly reflect compositional shifts across sectors
   rather than within-sector price/size variation. Kiva's raw data classify every loan into one of 15
   sectors (Agriculture, Food, Retail, Services, ... ), so `C(sector_name)` fixed effects are directly
   available and are added without needing any additional external data.

```python
loan_2c_path = Path("outputs/final_panel_loan_level_augmented_2c.csv")
loan_2c = pd.read_csv(loan_2c_path)
print("Loaded augmented loan-level file (adds sector_name, gdp_pc_const2015 in levels):", loan_2c_path)
print("Shape:", loan_2c.shape)

loan_2c["c_log_gdp_pc"] = loan_2c["log_gdp_pc"] - loan_2c["log_gdp_pc"].mean()
loan_2c["c_log_gdp_pc_sq"] = loan_2c["c_log_gdp_pc"] ** 2

us_share = (loan_2c["country_code"] == "US").mean()
print(f"US share of loans in this file: {us_share:.2%}")

lf6_sector = "log_loan_amount ~ c_log_gdp_pc + c_log_gdp_pc_sq + institutional_pca1 + log_population + C(year) + C(sector_name)"
lf6_rel = "log_relative_loan_size ~ c_log_gdp_pc + c_log_gdp_pc_sq + institutional_pca1 + log_population + C(year)"

no_us_data = loan_2c[loan_2c["country_code"] != "US"].copy()
lm6_no_us = fit_ols(lf6, no_us_data, se="cluster", cluster_col="country_year_id")
lm6_rel = fit_ols(lf6_rel, loan_2c, se="cluster", cluster_col="country_year_id")
lm6_sector = fit_ols(lf6_sector, loan_2c, se="cluster", cluster_col="country_year_id")

# Reference column: lm6 re-estimated on this same augmented file, to confirm it matches Section 5.2's lm6
lm6_on_2c_file = fit_ols(lf6, loan_2c, se="cluster", cluster_col="country_year_id")

robustness_2c_table = pd.DataFrame({
    "(1) lm6 baseline": {
        "c_log_gdp_pc": lm6_on_2c_file.params["c_log_gdp_pc"], "p": lm6_on_2c_file.pvalues["c_log_gdp_pc"],
        "c_log_gdp_pc_sq": lm6_on_2c_file.params["c_log_gdp_pc_sq"], "p (sq)": lm6_on_2c_file.pvalues["c_log_gdp_pc_sq"],
        "N": int(lm6_on_2c_file.nobs),
    },
    "(2) Excl. US": {
        "c_log_gdp_pc": lm6_no_us.params["c_log_gdp_pc"], "p": lm6_no_us.pvalues["c_log_gdp_pc"],
        "c_log_gdp_pc_sq": lm6_no_us.params["c_log_gdp_pc_sq"], "p (sq)": lm6_no_us.pvalues["c_log_gdp_pc_sq"],
        "N": int(lm6_no_us.nobs),
    },
    "(3) Relative size (loan/GDPpc)": {
        "c_log_gdp_pc": lm6_rel.params["c_log_gdp_pc"], "p": lm6_rel.pvalues["c_log_gdp_pc"],
        "c_log_gdp_pc_sq": lm6_rel.params["c_log_gdp_pc_sq"], "p (sq)": lm6_rel.pvalues["c_log_gdp_pc_sq"],
        "N": int(lm6_rel.nobs),
    },
    "(4) Sector FE": {
        "c_log_gdp_pc": lm6_sector.params["c_log_gdp_pc"], "p": lm6_sector.pvalues["c_log_gdp_pc"],
        "c_log_gdp_pc_sq": lm6_sector.params["c_log_gdp_pc_sq"], "p (sq)": lm6_sector.pvalues["c_log_gdp_pc_sq"],
        "N": int(lm6_sector.nobs),
    },
}).T

print()
print("Table 12. Loan-Level Robustness: lm6 Under Four Specifications")
display(robustness_2c_table.round(4))
```

```python
# Important caveat on column (3): log(loan_amount / gdp_pc_const2015) = log_loan_amount - log_gdp_pc,
# and log_gdp_pc is already included (in centered form) as the model's own linear regressor. By the
# Frisch-Waugh-Lovell theorem, subtracting a regressor that is already in the design matrix from the
# dependent variable only shifts THAT regressor's own coefficient (by exactly -1) and leaves every other
# coefficient -- including the quadratic term -- numerically unchanged. This is a mechanical identity, not
# an empirical finding, so column (3) as specified cannot actually test whether the U-shape is a
# nominal-price-level artifact.
print("FWL check: column (1) linear coef. - 1 =", round(lm6_on_2c_file.params['c_log_gdp_pc'] - 1, 6),
      "vs. column (3) linear coef. =", round(lm6_rel.params['c_log_gdp_pc'], 6))
print("Column (1) quadratic coef. =", round(lm6_on_2c_file.params['c_log_gdp_pc_sq'], 6),
      "vs. column (3) quadratic coef. =", round(lm6_rel.params['c_log_gdp_pc_sq'], 6), "(identical by construction)")

# A genuinely informative version of the same idea: drop the linear log-GDP control (whose presence is
# exactly what causes the mechanical cancellation above) and test whether curvature in the DEFLATED
# measure survives on its own.
lf6_rel_no_linear = "log_relative_loan_size ~ c_log_gdp_pc_sq + institutional_pca1 + log_population + C(year)"
lm6_rel_no_linear = fit_ols(lf6_rel_no_linear, loan_2c, se="cluster", cluster_col="country_year_id")
print()
print("Supplementary check -- relative size, WITHOUT the (self-cancelling) linear GDP control:")
print("  quadratic coef. =", round(lm6_rel_no_linear.params["c_log_gdp_pc_sq"], 4),
      "| p =", round(lm6_rel_no_linear.pvalues["c_log_gdp_pc_sq"], 4), "| N =", int(lm6_rel_no_linear.nobs))
```

```python
# Lind-Mehlum U-test applied to each of the four Table 12 variants (reusing the function from Section 5.4)
x_min_2c, x_max_2c = loan_2c["c_log_gdp_pc"].min(), loan_2c["c_log_gdp_pc"].max()
x_min_no_us, x_max_no_us = no_us_data["c_log_gdp_pc"].min(), no_us_data["c_log_gdp_pc"].max()

utest_variants = {
    "(1) lm6 baseline": lind_mehlum_utest(lm6_on_2c_file, "c_log_gdp_pc", "c_log_gdp_pc_sq", x_min_2c, x_max_2c),
    "(2) Excl. US": lind_mehlum_utest(lm6_no_us, "c_log_gdp_pc", "c_log_gdp_pc_sq", x_min_no_us, x_max_no_us),
    "(3) Relative size": lind_mehlum_utest(lm6_rel, "c_log_gdp_pc", "c_log_gdp_pc_sq", x_min_2c, x_max_2c),
    "(4) Sector FE": lind_mehlum_utest(lm6_sector, "c_log_gdp_pc", "c_log_gdp_pc_sq", x_min_2c, x_max_2c),
}
utest_variants_table = pd.DataFrame(utest_variants).T[["t_low", "t_high", "u_test_p", "shape_confirmed"]]
print("Table 13. Lind-Mehlum U-Test Applied to All Four Loan-Level Robustness Variants")
display(utest_variants_table.round(4))
```

Table 12 and Table 13 together show that the loan-level U-shape is not an artifact of any single
specification choice, **with one important caveat that must be stated plainly rather than buried**.

**Excluding the US (column 2)** barely matters and if anything strengthens the pattern: the US is only
0.9% of loans in this sample, and the quadratic term *rises* slightly to 0.332 (still p < 0.001) once it is
dropped, so the "richer countries get larger loans again" half of the U-shape is clearly not a US-specific
artifact.

**Relative loan size (column 3) requires a genuine caveat.** As the code comments above show directly, the
quadratic coefficient in column 3 is **numerically identical** to column 1's (0.2447 in both, to four
decimal places) — not approximately similar, but exactly the same, because `log(loan_amount/gdp_pc)` is
algebraically `log_loan_amount - log_gdp_pc`, and `log_gdp_pc` is already the model's own linear
regressor. By the Frisch-Waugh-Lovell theorem, subtracting a variable that is already in the regression
from the dependent variable can only change *that variable's own* coefficient (here, the linear term drops
by exactly 1, from 0.206 to −0.794) and mechanically **cannot** move any other coefficient, including the
quadratic term. **This means column 3, exactly as specified, is not actually a test of whether the
nominal-dollar U-shape survives deflating by local price levels — it is guaranteed to reproduce the
original quadratic term regardless of the true data-generating process.** This is exactly the kind of
result that should be reported plainly rather than presented as if it were independent confirmation: it
looks like a robustness check that "passed," but it could not have failed. The supplementary check computed
above — the same relative-size outcome, but with the self-cancelling linear control removed — is a
genuinely different regression, and there the quadratic term **attenuates but survives**: 0.148 (down from
0.245) and still significant at p = 0.002. The honest reading is therefore that **some, but probably not
all, of the nominal-dollar U-shape's magnitude reflects local price/income-level scaling**, while a
statistically significant U-shaped pattern in the deflated measure remains.

**Sector fixed effects (column 4)** leave the quadratic term essentially unchanged (0.234 vs. 0.245 in the
baseline, both p < 0.001), so compositional shifts across Kiva's 15 lending sectors are not driving the
result — the U-shape holds up *within* sectors, not just across them.

Table 13 confirms with the formal Lind–Mehlum test that all four variants pass decisively (all
one-sided p ≈ 0, both end-slopes individually significant with the correct opposite signs in every
case) — including column 3, for the mechanical reason just explained, which is precisely why Table 13's
"pass" for column 3 should not be read as additional independent evidence beyond what the supplementary,
linear-control-free check already provides.

## 7. Limitations

**Sample selection (partially addressed).** The country-year and loan-level panels (Sections 5.1–5.2) cover
only the countries in which Kiva was already active between 2013 and 2017; whether Kiva enters a country at
all is itself a decision shaped by local infrastructure, regulation, and partner-organization availability,
and that entry decision is not modeled in those two panels. Section 5.3 directly addresses part of this by
building a global panel that includes zero-lending country-years and modeling entry with a Logit, but it
still cannot say *why* Kiva enters some countries and not others beyond the reduced-form relationship with
GDP per capita and population — the entry decision itself (partner-organization availability, regulatory
approval, etc.) is not observed or modeled structurally.

**Log transform and the excluded extensive margin (addressed at the country level, not the loan level).**
`log_total_loan_amount` and `log_loan_amount` are well-defined and strictly positive for every row in their
respective panels (there are no zero-loan country-years or zero-dollar loans within the Kiva-only samples).
Section 5.3's global panel resolves this at the country level by including true zeros and modeling them
with a Logit rather than dropping them. There is no loan-level analogue of this concern, since individual
realized loans are by definition non-zero.

**Short panel.** The panel spans only five years (2013–2017). Some of the observed variation may reflect
the global expansion of the Kiva platform over this period rather than only structural differences across
countries; Section 2.3 checks that 2013 is not a partial/incomplete year and Section 6.2 confirms the main
results are unchanged if 2013 is dropped, but the panel is still short relative to typical macro panels, and
year fixed effects (included in most specifications) only partially absorb platform-growth effects that may
themselves vary by country.

**Price-level and currency-conversion effects (partially addressed).** Section 6.3 shows that a naive test
of whether the loan-level U-shape survives deflating by local GDP per capita is mechanically guaranteed to
reproduce the original result (a Frisch-Waugh-Lovell identity, not new evidence), and that a corrected
version of the same test shows the U-shape attenuates (from 0.245 to 0.148) but remains significant. This
suggests part, but apparently not all, of the nominal-dollar loan-size U-shape reflects local price-level
scaling rather than a real change in relative loan size — a distinction future work could pursue further
with a proper PPP or local-currency deflator rather than GDP per capita alone.

**Descriptive, not causal.** All specifications are estimated by OLS, Logit, or PPML on observational data.
Development, institutional quality, and Kiva lending are jointly determined by many unobserved factors, so
the coefficients reported here should be read as conditional associations, not treatment effects.

## 8. Conclusion

This paper set out to test whether Kiva microfinance lending activity follows an inverted-U relationship
with economic development, peaking in lower-middle income countries. The country-year evidence does not
support that hypothesis: the quadratic term in Table 4 is negative but statistically indistinguishable from
zero in all four specifications (p > 0.10), and the linear component is negative on average, so the more
defensible reading is that **total lending volume tends to decline with development** across most of the
sample, without strong evidence of a rise-then-fall shape. Section 5.4's formal Lind–Mehlum (2010) U-test
confirms this is not just a significance-testing artifact: the slope at the low end of the sample is not
even individually significant (t ≈ 1.08), and the Fieller confidence interval for the implied turning point
is unbounded.

At the loan level, a different and much better-supported pattern emerges: the quadratic term in Tables 6–7
is **positive and significant at the 1% level**, implying that **individual loan size follows a U-shape**
— falling and then rising with development — the opposite sign from the (insignificant) country-level
curvature. The Lind–Mehlum test confirms this shape decisively (both end-slopes individually significant,
opposite-signed, one-sided p < 0.001), with a Fieller turning-point interval of roughly \$1,150–\$1,970
that lies comfortably inside the sample range. Section 6.3 shows this survives excluding the US, sector
fixed effects, and (with attenuated but still significant magnitude once a mechanical cancellation is
correctly accounted for) deflating by local price levels.

Section 5.3 adds a third margin that resolves an ambiguity in the first two: because
`final_panel_country_year.csv` only contains country-years where Kiva was already active, Table 4 cannot
distinguish "no nonlinearity anywhere" from "the nonlinearity is in whether Kiva enters a country at all."
Building a global panel that includes true zero-lending country-years and estimating a Logit for
Kiva's extensive margin shows **significant concavity with a steep decline**: the probability that
Kiva is active in a country is highest among the poorest deciles (peaking in the raw data at the
second decile) and falls to near zero among the richest countries (Table 9, Figure 8), though the
initial rise at the very bottom fails the formal low-end slope test and so a completed inverted-U is
not established (Table 10). PPML on total lending volume — both including zeros and
restricted to Kiva-active country-years — shows no such pattern, consistent with the OLS-in-logs result.

Because `log_total_loan_amount`, `kiva_active`, and `log_loan_amount` measure three different objects —
aggregate country-level volume, whether Kiva operates in a country at all, and the size of a single
realized loan — these are not competing estimates of one underlying relationship. They are three distinct
findings about three different margins of microfinance activity:

1. **Extensive margin (country level, whether Kiva is active at all):** significantly concave —
   entry probability declines steeply with development; the raw-data peak in the second income
   decile is suggestive, but the initial rise is not statistically confirmed by the Lind–Mehlum test
   (Sections 5.3–5.4, Tables 9–10).
2. **Quantity margin (country level, conditional on Kiva being active):** not robustly nonlinear in
   development; if anything it declines as countries get richer within this sample, and fails the formal
   U-test (Section 5.1, 5.4).
3. **Size margin (loan level):** robustly and significantly U-shaped in development, confirmed by the
   formal U-test and by four separate robustness variants (Sections 5.2, 5.4, 6.3).
4. **These are not two (or three) tests of the same claim.** A country can be less likely to receive any
   Kiva lending as it develops, while conditional on receiving some, the total volume and the typical loan
   size follow different — and in one case opposite-signed — patterns. Treating any one significant
   coefficient as "support" for another margin's insignificant one would conflate different dependent
   variables measuring different economic objects.

Both results are conditional on the sample-selection and short-panel limitations discussed in Section 7,
and both are descriptive rather than causal. The extensive-margin analysis in Section 5.3 addresses part of
the original "excluded zeros" limitation directly, but the loan-level findings remain conditional on a loan
having been made. Future work could usefully model Kiva's entry decision more richly (e.g., with
partner-organization availability as a covariate), extend the panel beyond five years, and examine
borrower-level heterogeneity within the loan-level file to test whether the size U-shape reflects
compositional shifts in borrower type or loan purpose as countries develop, beyond the sector and
price-level checks already performed here.
